### Import Library and data

```python
import pandas as pd
import numpy as np

df = pd.read_csv(
    "../data/processed/tourism_arrivals_clean.csv"
)
```

### describe

```python
df["Tourist_Arrivals"].describe()
```

```python
mean_arrivals = df["Tourist_Arrivals"].mean()

print("Mean:", mean_arrivals)
```

```python
median_arrivals = df["Tourist_Arrivals"].median()

print("Median:", median_arrivals)
```

```python
std_arrivals = df["Tourist_Arrivals"].std()

print("Standard Deviation:", std_arrivals)
```

```python
print("Minimum:", df["Tourist_Arrivals"].min())
print("Maximum:", df["Tourist_Arrivals"].max())
```

### Country statistics

```python
country_stats = (
    df.groupby("Country")["Tourist_Arrivals"]
    .agg(
        Total="sum",
        Average="mean",
        Median="median",
        Maximum="max",
        Minimum="min"
    )
    .sort_values("Total", ascending=False)
)

country_stats.head(20)
```

### Year statistics

```python
year_stats = (
    df.groupby("Year")["Tourist_Arrivals"]
    .agg([
        "sum",
        "mean",
        "median",
        "std",
        "min",
        "max"
    ])
)

year_stats
```

```python
year_stats.to_csv(
    "../data/processed/yearly_descriptive_statistics.csv"
)

country_stats.to_csv(
    "../data/processed/country_descriptive_statistics.csv"
)
```
